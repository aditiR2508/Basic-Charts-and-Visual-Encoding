```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data1 = {
    'Days': ['Mon','Tue','Wed','Thu','Fri'],
    'Study_Hours': [2,3,2.5,4,3.5],
    'Marks': [50,60,55,70,65],
    'Attendance': [60,70,65,80,75],
    'Sleep_Hours': [6,7,6,8,7],
    'Assignments_Completed': [1,2,1,3,2]
}
df = pd.DataFrame(data1)
print(df)
```

```python
# 1. Line Chart (Trend)

plt.plot(df['Days'], df['Study_Hours'], marker='o') ## marker: o-->circle, s-->square, *-->star, ^-->triangle, X-->cross

plt.title("Study Hours Trend")
plt.xlabel("Days")
plt.ylabel("Study Hours")

plt.show()
```

# Other Common Marker Types Meaning

## 'o'--->Circle

## 's'--->Square

## '^'--->Triangle

## 'x'--->Cross

## '*'--->Star

# Common Color Options

## 1. Using names:

## 'blue'

## 'red'

## 'green'

## 'black'

## 'orange'

## 2. Short forms:

## 'b' → blue

## 'r' → red

## 'g' → green

## 'k' → black


```python
# Advanced line chart
plt.figure(figsize=(7,4))

plt.plot(df['Days'], df['Study_Hours'], marker='o', color='red', label='Study Hours')

plt.plot(df['Days'], df['Marks'], marker='*', color='green', label='Marks')

plt.title("Study Hours and Marks Trend")
plt.xlabel("Days")
plt.ylabel("Values")
plt.legend()

plt.show()
```

```python
# 2. Bar Chart (Comparison)

plt.bar(df['Days'], df['Marks'], color='skyblue')

plt.title("Marks Comparison")
plt.xlabel("Days")
plt.ylabel("Marks")

plt.show()
```

```python
# Advanced Bar graph
bars = plt.bar(df['Days'], df['Marks'], color='yellow')
for bar in bars:
  y = bar.get_height()
  plt.text(bar.get_x() + bar.get_width()/2, y, str(y), ha='center', va='bottom')

plt.title("Marks Comparison")
plt.xlabel("Days")
plt.ylabel("Marks")

plt.grid(axis='y') ## adds horizontal grid
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(len(df['Days']))   # positions for bars
width = 0.35                     # width of each bar

plt.bar(x - width/2, df['Study_Hours'], width, label='Study Hours', color='r')
plt.bar(x + width/2, df['Marks'], width, label='Marks', color='b')

plt.xticks(x, df['Days'])

plt.title("Study Hours vs Marks")
plt.xlabel("Days")
plt.ylabel("Values")
plt.legend()

plt.show()
```

```python
# 3. Histogram (Distribution)
##shows how values are spread
data2 = {
    'Marks': [50,60,55,70,65,75,80,85,90,45,68,72,77,82,88,90,95,98,105,108]
}
df = pd.DataFrame(data2)
plt.hist(df['Marks'], bins=7, edgecolor='black', alpha=0.5) ## alpha lies between 1 and 0, 1 being darkest and 0 lightest
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
# Advanced Histogram(using mean)
import matplotlib.pyplot as plt
import numpy as np

# Example data (more values for better histogram)
Marks = [50, 60, 55, 70, 65, 75, 80, 85, 90, 45, 68, 72, 77, 82, 88]

# Calculate mean
mean_value = np.mean(Marks)

plt.figure(figsize=(7,4))

plt.hist(Marks, bins=6, color='pink', edgecolor='blue')

# Mean line
plt.axvline(mean_value, color='red', linestyle='--', linewidth=2, label='Mean')

plt.title("Advanced Histogram of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.legend()

plt.grid(linestyle='--', alpha=0.5)

plt.show()
```

```python
# 4. Scatter Plot (Relationship)
import matplotlib.pyplot as plt

plt.scatter(df['Study_Hours'], df['Marks'], marker='*', color='magenta')

plt.title("Study Hours Vs. Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
# Conditional Scatter plot
## Example category

df['Result'] = ['Fail','Pass','Fail','Pass','Pass']
colors = ['red' if r=='Fail' else 'green' for r in df['Result']]
plt.scatter(df['Study_Hours'], df['Marks'], c=colors, marker='*')

plt.title("Study Hours Vs. Marks(Result-Wise)")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
# Dataset 2
data = {
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun', 'Mon', 'Tue', 'Wed'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': [200, 220, 250, 270, 300, 350, 400, 280, 260, 290],
    'Profit': [20, 25, 30, 35, 50, 60, 80, 40, 38, 45],
    'Customers': [30, 35, 40, 45, 50, 65, 80, 55, 48, 52],
    'Category': ['Food', 'Clothing', 'Electronics', 'Food', 'Clothing', 'Electronics', 'Food', 'Clothing', 'Electronics', 'Food']
}

df = pd.DataFrame(data)
print(df)
```
